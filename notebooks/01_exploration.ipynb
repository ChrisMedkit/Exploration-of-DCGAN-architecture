## 2. Class balance
Images per class: is the data balanced enough that the GAN won't only learn the big classes?

```python
counts = torch.bincount(labels)
print(counts.min().item(), counts.median().item(), counts.max().item())
plt.bar(range(102), counts.sort(descending=True).values); plt.xlabel('class (sorted)'); plt.ylabel('images'); plt.show()
```
